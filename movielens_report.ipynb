**Резкий пик в 2000 году - 10 061 оценка, при том что соседние 1999 и 2001 годы дают лишь 2 439 и 3 922**. Это больше похоже на разовую пакетную загрузку исторических оценок (или технический артефакт при формировании датасета), чем на органический всплеск интереса пользователей. Более поздние пики (2007, 2015–2017) уже выглядят правдоподобнее и, вероятно, связаны с ростом самого сервиса и мобильных клиентов.

## **V. Обогащение через TMDB: бюджеты, сборы, режиссёры**

**Для ускорения работы с TMDB API анализ в этом разделе выполнен на подвыборке из 30 фильмов - первых по movieId в датасете. Поскольку movieId в MovieLens в целом коррелирует с хронологией добавления фильмов в каталог, эта выборка смещена в сторону более старых релизов и не является случайной или репрезентативной для всего датасета.*

*Логика подсчёта и сам код работают корректно и не требуют изменений - ограничение носит исключительно характер выборки данных, а не ошибку в реализации.*

***Приведённые ниже цифры стоит рассматривать как иллюстрацию работы пайплайна, а не как итоговые выводы о полном датасете MovieLens.***

```python
%timeit Links(LINKS_PATH, MOVIES_PATH, TMDB_API_KEY)
links = Links(LINKS_PATH, MOVIES_PATH, TMDB_API_KEY)
links.movies_data
```

```python
links = Links("../datasets/links.csv", "../datasets/movies.csv", TMDB_API_KEY)

top_30_ids = set(list(links.movies_data.keys())[:30])

links.movies_data = {
    movie_id: info
    for movie_id, info in links.movies_data.items()
    if movie_id in top_30_ids
}
```

### Общая информация о фильме

Для заданного списка movieId метод тянет данные из TMDB и собирает список по запрошенным полям. Сортировка по movieId по убыванию.

```python
%timeit links.get_imdb([1, 2, 3], ["Title", "Director", "Budget", "Runtime"])
links.get_imdb([1, 2, 3], ["Title", "Director", "Budget", "Runtime"])
```

### Наиболее дорогие фильмы

Возвращает top-n {название: бюджет} по всем фильмам в self.movies_data, отсортированных по убыванию бюджета.

```python
links.most_expensive(3)
```

### Наиболее продуктивные режиссёры

Проходит по всем фильмам в self.movies_data, считает, у скольких фильмов каждый режиссёр значится как Director, возвращает top-n {режиссёр: количество фильмов}.

```python
%timeit links.top_directors(3)
links.top_directors(3)
```

### Фильмы с самой дорогой минутой

Возвращает top-n {название: бюджет / хронометраж}, отсортировано по убыванию - учитываются только фильмы с ненулевыми бюджетом и хронометражом.

```python
%timeit links.top_cost_per_minute(3)
links.top_cost_per_minute(3)
```

### Наиболее длинные фильмы

Возвращает top-n {название: хронометраж} по убыванию длительности.

```python
%timeit links.longest(3)
links.longest(3)
```

### Самые окупаемые фильмы

Возвращает top-n {название: прибыль}, где прибыль = выручка - бюджет, отсортировано по убыванию.
